```python
def remove_last(lst):
    lst.pop()

my_list = [10, 20, 30, 40]

print("Before function:", my_list)

remove_last(my_list)

print("After function:", my_list)
```

```python
def change_string(s):
    s = "X" + s[1:]
    print("Inside function:", s)

my_string = "Hello"

print("Before function:", my_string)

change_string(my_string)

print("After function:", my_string)
```

```python
def add_entry(d):
    d["age"] = 20

my_dict = {"name": "Rahul"}

print("Before:", my_dict)

add_entry(my_dict)

print("After add_entry:", my_dict)
```

```python
def factorial(n):
    result = 1

    for i in range(1, n + 1):
        result = result * i

    return result

n = int(input("Enter a number: "))

print("Factorial:", factorial(n))
```

```python
def maximum(a, b):
    if a > b:
        return a
    else:
        return b

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

result = maximum(num1, num2)

print("Greater number:", result)
```

```python
def celsius_to_fahrenheit(c):
    return (c * 9 / 5) + 32

celsius = float(input("Enter temperature in Celsius: "))

fahrenheit = celsius_to_fahrenheit(celsius)

print("Temperature in Fahrenheit:", fahrenheit)
```
